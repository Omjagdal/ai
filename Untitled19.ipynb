```python
import tensorflow as tf

# Tensor creation
a = tf.constant([[1, 2], [3, 4]])
b = tf.constant([[5, 6], [7, 8]])

print("Tensor A:\n", a)
print("Tensor B:\n", b)

# Basic operations
print("\nAddition:\n", a + b)
print("\nMultiplication:\n", a * b)
print("\nMatrix Multiplication:\n", tf.matmul(a, b))

# Tensor manipulation
print("\nReshaped Tensor:", tf.reshape(a, [4]))
print("Transpose:\n", tf.transpose(a))

# Eager execution
x = tf.constant(5)
y = tf.constant(10)
z = x * y + 2

print("\nEager Execution Result:", z.numpy())
print("TensorFlow Version:", tf.__version__)
```

```python
import seaborn as sns
import pandas as pd
from sklearn.preprocessing import MinMaxScaler, LabelEncoder

# Load built-in dataset
df = sns.load_dataset("titanic")

print(df.head())

# Select a few columns
df = df[["age", "fare", "sex", "class"]]

# Handle missing values
df["age"] = df["age"].fillna(df["age"].mean())
df["fare"] = df["fare"].fillna(df["fare"].mean())

# Normalize numerical columns
scaler = MinMaxScaler()
df[["age", "fare"]] = scaler.fit_transform(df[["age", "fare"]])

# Encode categorical columns
encoder = LabelEncoder()
df["sex"] = encoder.fit_transform(df["sex"])
df["class"] = encoder.fit_transform(df["class"])

print("\nProcessed Dataset:")
print(df.head())
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

# Load dataset
df = sns.load_dataset("titanic")

# Select numerical columns
data = df[["age", "fare", "pclass"]].dropna()

# Histogram
plt.figure(figsize=(6, 4))
plt.hist(data["age"], bins=20)
plt.title("Age Distribution")
plt.xlabel("Age")
plt.ylabel("Frequency")
plt.show()

# Scatter plot
plt.figure(figsize=(6, 4))
plt.scatter(data["age"], data["fare"])
plt.title("Age vs Fare")
plt.xlabel("Age")
plt.ylabel("Fare")
plt.show()

# Correlation heatmap
plt.figure(figsize=(6, 4))
sns.heatmap(data.corr(), annot=True)
plt.title("Correlation Heatmap")
plt.show()
```

```python
!pip install gensim faiss-cpu
from gensim.models import Word2Vec
import faiss
import numpy as np

# Small text corpus
sentences = [
    ["machine", "learning", "is", "interesting"],
    ["deep", "learning", "uses", "neural", "networks"],
    ["artificial", "intelligence", "uses", "machine", "learning"],
    ["data", "science", "uses", "data"]
]

# Train Word2Vec
model = Word2Vec(
    sentences,
    vector_size=50,
    window=3,
    min_count=1,
    workers=1
)

# Get all words
words = list(model.wv.index_to_key)

# Create embedding vectors
vectors = np.array([model.wv[word] for word in words]).astype("float32")

# Create FAISS vector database
index = faiss.IndexFlatL2(50)
index.add(vectors)

# Query word
query = "learning"
query_vector = np.array([model.wv[query]]).astype("float32")

# Search top 3 similar words
distances, indices = index.search(query_vector, 3)

print("Query:", query)

print("\nSimilar Words:")
for i in indices[0]:
    print(words[i])

# Word2Vec semantic similarity
print("\nSemantic Similarity:")
print(model.wv.most_similar("learning", topn=3))
```

```python
import tensorflow as tf
import numpy as np
import matplotlib.pyplot as plt

# Load MNIST dataset
(x_train, _), (_, _) = tf.keras.datasets.mnist.load_data()

# Use 5000 images
x_train = x_train[:5000].astype("float32") / 255.0
x_train = x_train[..., np.newaxis]

# Add less noise
noise = np.random.normal(0, 0.3, x_train.shape)
noisy_images = np.clip(x_train + noise, 0, 1)

# Simple denoising model
model = tf.keras.Sequential([
    tf.keras.layers.Input(shape=(28, 28, 1)),

    tf.keras.layers.Conv2D(32, 3, activation="relu", padding="same"),
    tf.keras.layers.Conv2D(64, 3, activation="relu", padding="same"),
    tf.keras.layers.Conv2D(32, 3, activation="relu", padding="same"),

    tf.keras.layers.Conv2D(1, 3, activation="sigmoid", padding="same")
])

model.compile(
    optimizer="adam",
    loss="mse"
)

# Train
model.fit(
    noisy_images,
    x_train,
    epochs=10,
    batch_size=64
)

# Generate cleaned images
generated_images = model.predict(noisy_images[:5])

# Display results
plt.figure(figsize=(12, 7))

for i in range(5):

    plt.subplot(3, 5, i + 1)
    plt.imshow(x_train[i].squeeze(), cmap="gray")
    plt.title("Original")
    plt.axis("off")

    plt.subplot(3, 5, i + 6)
    plt.imshow(noisy_images[i].squeeze(), cmap="gray")
    plt.title("Noisy")
    plt.axis("off")

    plt.subplot(3, 5, i + 11)
    plt.imshow(generated_images[i].squeeze(), cmap="gray")
    plt.title("Generated")
    plt.axis("off")

plt.tight_layout()
plt.show()
```
